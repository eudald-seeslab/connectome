```python
import os
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import matplotlib.image as mpimg

from paths import PROJECT_ROOT

plots_dir = os.path.join(PROJECT_ROOT, "plots")

# Create the composite figure
fig = plt.figure(figsize=(4, 10), dpi=300)
gs = GridSpec(1, 2, figure=fig, wspace=0., hspace=0.1)

# Define subplot areas
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])

# Load the images
dir_path = os.path.join(PROJECT_ROOT, "plots")
ax1.imshow(mpimg.imread(os.path.join(dir_path, "efficiency_comparison.png")))
ax2.imshow(mpimg.imread(os.path.join(dir_path, "accuracy_comparison.png")))

for ax in [ax1, ax2]:
    ax.axis("off")

# Add panel labels with improved positioning
label_params = {"fontsize": 8, "fontweight": "bold", "va": "top", "ha": "left"}

ax1.text(0, 1.05, "a", transform=ax1.transAxes, **label_params)
ax2.text(0, 1.05, "b", transform=ax2.transAxes, **label_params)

plt.tight_layout()

# save
plt.savefig(
    os.path.join(plots_dir, "efficiency_panel.png"),
    dpi=300,
    bbox_inches="tight",
)
plt.savefig(
    os.path.join(plots_dir, "efficiency_panel.pdf"), bbox_inches="tight"
)

plt.show()
```

```python
from matplotlib.gridspec import GridSpec
import matplotlib.image as mpimg

# Create the composite figure
fig = plt.figure(figsize=(12, 12), dpi=300)
gs = GridSpec(1, 3, figure=fig, wspace=0., hspace=0.1)

# Define subplot areas
## Upper row
ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax3 = fig.add_subplot(gs[0, 2])

# Load the images
dir_path = os.path.join(PROJECT_ROOT, "plots")
ax1.imshow(mpimg.imread(os.path.join(dir_path, "neuron_activation.png")))
ax2.imshow(mpimg.imread(os.path.join(dir_path, "thinking_neuron_activation.png")))
ax3.imshow(mpimg.imread(os.path.join(dir_path, "propagation_distance.png")))

for ax in [ax1, ax2, ax3]:
    ax.axis("off")

# Add panel labels with improved positioning
label_params = {"fontsize": 20, "fontweight": "bold", "va": "top", "ha": "left"}

ax1.text(0, 1.05, "a", transform=ax1.transAxes, **label_params)
ax2.text(0, 1.05, "b", transform=ax2.transAxes, **label_params)
ax3.text(0, 1.05, "c", transform=ax3.transAxes, **label_params)

plt.tight_layout()

# save
plt.savefig(
    os.path.join(plots_dir, "neuron_activation_panel.png"),
    dpi=300,
    bbox_inches="tight",
)
plt.savefig(
    os.path.join(plots_dir, "neuron_activation_panel.pdf"), bbox_inches="tight"
)

plt.show()
```
